# Visualização de Dados no Python #108



```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
dados = {
    'X' : [1,2,3,4,5,6],
    'Y' : [120,110,130,145,118,125],
    'Z' : [95,54,86,77,90,81],
}

base = pd.DataFrame(dados)
```

```python
base.head()
```

## Criando um plot

```python
plt.style.use('_mpl-gallery')

# make data
i = base.index
x = base['X']
y = base['Y']
z = base['Z']


# plot
fig, ax = plt.subplots()

ax.plot(i, x+100, markeredgewidth=2)
# ax.plot(i, y, 'x', markeredgewidth=2)
ax.plot(i, y, linewidth=2.0)
ax.plot(i, z, 'o-',linewidth=2)

plt.show()
```

## Separando e 2 graficos um em baixo do outro

```python
fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(7, 3))
ax[0].plot(x, y, linewidth=2.0)

ax[0].set(xlim=(x[0], x[len(x) - 1]),
       ylim=(60, 150))

ax[1].plot(x, z, linewidth=2.0)

ax[1].set(xlim=(x[0], x[len(x) - 1]),
       ylim=(50, 150))

plt.show()
```

```python
## Melhorando a Visualização

### figsize, gridspec_kw
fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(7, 3), gridspec_kw={'height_ratios': [3,1]})
w=0.4
ax[0].bar(x, y, linewidth=2.0, color='red', width=w)
ax[0].bar(x-w/2 , z, linewidth=2.0, width=w)

# ax[0].set(xlim=(x[0], x[len(x) - 1]),
#        ylim=(60, 150))

ax[1].bar(x, z, linewidth=2.0)

ax[1].set(xlim=(x[0], x[len(x) - 1]),
       ylim=(50, 150))
```

```python
## Melhorando a Visualização

### figsize, gridspec_kw
fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(7, 3), gridspec_kw={'height_ratios': [3,1]})
w=0.4
ax[0].bar(x, y, linewidth=2.0, color='red', width=w)
ax[0].bar(x-w/2 , z, linewidth=2.0, width=w)

# ax[0].set(xlim=(x[0], x[len(x) - 1]),
#        ylim=(60, 150))

plt.show()
```

## Annotate - Adicionando rotulos nos dados #F109

- para usar o annotate, vamos passar .annotate("<Texto>",<posicao(x,y)>)
- posicionar o annotate em cima da barra usando a posição y
- usar o parametro annotate para melhorar a visualização
    - ha -> alinhamento horizontal
    - xytext -> deslocamento x,y do texto em relação a um referencial
    - textcords -> referencial que vamos fazer o deslocamento acima
    - fontsize - tamanho da fonte
    - fontweight - colocando em negrito

## plot


```python
# plot 
fig, ax = plt.subplots(figsize=(7, 3))
w = 0.3

ax.bar(base.X[0:4]+w/2, base.Y[0:4], label="Y", width=w, color='red')
ax.bar(base.X[0:4]-w/2, base.Z[0:4], label="Z", width=w)

ax.bar(base.X[4:6]+w/2, base.Y[4:6], width=w, color='pink')
ax.bar(base.X[4:6]-w/2, base.Z[4:6], width=w, color="violet")


for i in range (0, 6):
    ax.annotate(base.Z[i],((base.X[i]-w/2, base.Z[i])), ha="center", xytext=(0,5) ,textcoords="offset points")
    ax.annotate(base.Y[i],((base.X[i]+w/2, base.Y[i])), ha="center", xytext=(0,5) ,textcoords="offset points")
    
    
ax.set_yticks([0,170])
ax.yaxis.set_visible(False)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)




plt.legend()
plt.show()

## Ajustando o eixo y para melhorar a visualização dos rótulos
### .set_yticks -> alterar o intervalo do eixo y
### .yaxis.set_visible permite que os valores do eixo y não fiquem visiveis


## Retirando bordas do gráfico
### Usando o .spines[].ser_visible(False) é possivel retirar as bordas de um gráfico


## Separando realizado de projetado
### Podemos cada gráfico em 2 gráficos diferentes para fazer o relizado e o projetado, mudando as cores das barras
### Azul (#C6F0F5) e Laranja (#FFC174)
```

```python
## Adicionando Retangulos no gráfico para destracar areas

fig, ax = plt.subplots()

ax1 = fig.add_axes([0.5-0.2/2, 0.1, 0.2, 0.5])
rect = ax1.patch
rect.set_facecolor('#E0E6DE')
rect.set_alpha(0.5)
ax1.xaxis.set_visible(False)
ax1.yaxis.set_visible(False)
ax1.spines['top'].set_visible(False)
ax1.spines['left'].set_visible(False)
ax1.spines['right'].set_visible(False)
ax1.spines['bottom'].set_visible(False)


plt.show()
```
