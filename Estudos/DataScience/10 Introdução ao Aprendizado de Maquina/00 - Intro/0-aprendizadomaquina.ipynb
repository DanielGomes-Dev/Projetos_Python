# Aprendizado de Máquina

- Podemos usar machine learning para conseguir chegar no mesmo resultado da equação abaixo:


```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

## Dados em uma reta


```python
dados = {
    'X': [1.3,2.5,3.8,5.1,6],
    'Y': [168,312,468,624,732]
}


dados = pd.DataFrame(dados)
```

**Plotando esses pontos em um gráfico**


```python
fig, ax = plt.subplots()
ax.scatter(dados.X, dados.Y)
plt.show()
```

**Existe uma reta que passa por esses pontos**

```python
fig, ax = plt.subplots()
ax.scatter(dados.X, dados.Y)
x=[dados.X[0], dados.X[4]]
y=[dados.Y[0], dados.Y[4]]
ax.plot(x,y)
plt.show()
```

**Essa reta é dada por**

f(x) = 12 + 120x


então para estimar oo valor de qualquer novo x bastar substituir nessa equação

se x = 4,5, qual será o valor de f(x);

f(x) = 12 + 120*4,5 = 552


### e  Se quisermos usar ML para traçar reta ?
- OBS: não vamos focar agorea no modelo de Machine Learning, somente vamos usar os coeficientes (linear e angular) para traçar a reta
- https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html




```python
from sklearn.linear_model import LinearRegression

modelo = LinearRegression()
modelo.fit(dados.X.values.reshape(-1,1), dados.Y)
```

```python
# Coeficiente linear

modelo.intercept_
```

```python
# Coeficiente angular

modelo.coef_
```

**Podemos primeiramente plotar os pontos em um gráfico**

```python
fig, ax = plt.subplots()
ax.scatter(dados.X, dados.Y)
ax.plot(x,y)
plt.show()
```

**Usar esses dois coeficientes para traçar a reta que melhor se adequa a esses pontos**

- para isso, podemos considerar um range de 1 a 6 (inclusive) para o x e aplicar a função dada por y = intercept* + x.coef* para traçar os valores de y


```python
x = np.arange(1,8)
y = modelo.intercept_ + x*modelo.coef_[0]

fig,ax = plt.subplots()

ax.scatter(dados.X, dados.Y)
ax.plot(x, y)
x2 = [dados.X[0], dados.X[4]]
y2 = [dados.Y[0], dados.Y[4]]
ax.plot(x2,y2)


```

**Observe que os valores do coeficiente angular e linear são exatamente iguais da reta sem usar machine learning, então substituindo na reta o resultado será o mesmo**

**Porém, em ML , podemos apenas usar o predict**

```python
# Prevendo algum valor usando o modelo que acabamos de criar

modelo.predict(pd.DataFrame([4.5]))

## Nesse caso o valor foi exatamente o mesmo
```

## Agora vamos supor que os dados não estão bem distribuido

```python
dados2 = {
    'X': [1.3,2.5,3.8,5.1,6],
    'Y': [168, 275,501,705,732]
}

dados2 = pd.DataFrame(dados2)
```

**Novamente plotando esses dados e a reta que passa pelo primeiro e pelo ultimo ponto**

```python
fig, ax = plt.subplots()
ax.scatter(dados2.X, dados2.Y)
x = [dados2.X[0], dados2.X[4]]
y = [dados2.Y[0], dados2.Y[4]]
ax.plot(x,y)
plt.show()
```

**E se usarmos ML para traçar a reta?**

```python
from sklearn.linear_model import LinearRegression
modelo = LinearRegression()
modelo.fit(dados2.X.values.reshape(-1, 1), dados2.Y)
```

**Sendo o coefiencia linear e angular:**

```python
# Coeficiente linear
print(modelo.intercept_)

# Coeficiente angular
print(modelo.coef_)
```

**Podemos traçar em vermelho a reta que foi dada pelo modelo**

```python
x_ML = np.arange(1,7)
y_ML = modelo.intercept_ + x_ML*modelo.coef_[0]

fig, ax = plt.subplots()

ax.scatter(dados2.X, dados2.Y)
x = [dados.X[0]+0.2, dados.X[4] - 0.3]
y = [dados.Y[0], dados.Y[4]]+8563/20,

ax.plot(x,y)
ax.plot(x_ML, y_ML, c="red")
```
