# Voltando as informações estatisticas

```python
# Selecionando a base que vamos mostrar informações estatisticas

display(base68.head())
base_estatisticas = base68[['Survived', 'Sex', 'Age']]

display(base_estatisticas.head())
display(base_estatisticas.value_counts())
display(base_estatisticas.info())

```

```python
# Media das colunas
base_estatisticas.mean()
```

```python
# contagem de registros
base_estatisticas.count()
```

```python
# Mediana

base_estatisticas.median()
```

```python
# Desvio padrão
base_estatisticas.std()
```

```python
# Resumo total de estatistico

base_estatisticas.describe()
```

```python
display(base_estatisticas[base_estatisticas['Age'] > 20])

# .loc e .iloc

# o .loc permite fazer a busca da mesma forma acima
display(base_estatisticas.loc[base_estatisticas['Age'] > 20])

# tambem perimite argumentos logicos

display(base_estatisticas.loc[(base_estatisticas['Age'] > 20) & (base_estatisticas['Age'] < 30) ])
```

```python
# loc permite filtras colunas de forma muito pratica

display(base_estatisticas.loc[(base_estatisticas['Age'] > 20) & (base_estatisticas['Age'] < 30),['Sex', 'Age'] ])
# display(base_estatisticas.loc[(base_estatisticas['Age'] > 20) & (base_estatisticas['Age'] < 30),['Sex', 'Age'] ].value_counts())

```

```python
# iloc vai usar o indice para filtrar

display(base_estatisticas.iloc[30:40])

# tambem serve para filtra colunas especificas

display(base_estatisticas.iloc[30:40, [0,2]])
```
