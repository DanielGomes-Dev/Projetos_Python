
# NumPy
- https://numpy.org/
- O nome NumPy vem de Numerical Python  - Biblioteca de computação numérica
    - Otimizada para cálculos pesados
- No NumPy vamos trabalhar com arrays, que são mais rápidos e mais performáticos que listas
    - Arrays são estruturas de dados que guardam itens do mesmo tipo (diferente das listas)
- "As operações matemáticas que devem ser executadas em arrays seriam extremamente ineficientes se os arrays não fossem homogêneos."
    - Assim como listas, são um conjunto de elementos ordenados que são mutáveis e de comprimento variável
    - Além disso, podemos fazer várias operações com arrays (como multiplicação e soma) que não podem ser feitas com listas mas são fundamentais para o nosso processo de Ciência de dados
    - Obs: listas são estrutudas de dados do Python e arrays não, array é uma estrutura de dados própria do numpy

```python
#Importar o numpy


import numpy as np
```

```python
# Venda e Comissao

venda_valor = [1500, 2500, 8000, 14200, 19000]
comissao = [5, 8, 8, 5, 12]

type(venda_valor)

# venda_valor / 100
# venda_valor * 100 Não funciona com lista
```

```python
# Transformando lista em array com numpy

# help(np.array)
venda_valor_array = np.array(venda_valor)
comissao_array = np.array(comissao)

print(comissao_array / 100);
print(venda_valor_array / 10)
print(venda_valor_array * 10)
print(venda_valor_array * (comissao_array/100))

```

```python
# F063 As propriedades de um array

## Copiando Exemplos

a = np.arange(20).reshape(4,5)
b = np.array([[1,2,3],[4,5,6]])

print('Type')

## Caso eu tenhqa um unico numero como texto todo o array vai ser considerado como texto

print(a)
print(type(a))

print(b)
print(type(b))

print(a.dtype)
print(b.dtype)



print('Dimensoes')

print(a.ndim) # nnumero de dimenssoes
print(b.ndim)

print('Linhas e Colunas')

print(a.shape) #(linha, coluna)
print(b.shape)
```

```python
# F064 Criando nossos próprios arrays (np.array, np.arange, np.linspace)


## Array de numeros inteiros
a64 = np.array([1,2,3,4,5])
print(a64)

## Array de float
b64 = np.array([1,2,3,4.7,5])
print(b64)

## Array de 0
zero64 = np.zeros(8)
um64 = np.ones(8)
zeroReshape64 = np.zeros(8).reshape(2,4)
print(zero64, um64, zeroReshape64);

zeroacem64 = np.arange(101)
print(zeroacem64)

arr15a25 = np.arange(15,26)
print(arr15a25)

arr15a25withpass = np.arange(15,26, 2)
print(arr15a25withpass)


## Criando array com os valores igualmente espaçados (utilizando o linspace)
linspace64 = np.linspace(5,10, 12)
print(linspace64, 'linspace64')

### Array de uma dimensão é o que chamamos de vetor e de duas dimensões de matriz
### Vamos considerar esses dados abaixo
dados_vendas64 = np.array( [[150000, 230000, 82000, 143000, 184000], [5,8,8,5,12]] )
print(dados_vendas64, 'matriz')
print(dados_vendas64.shape, 'matriz')
print(dados_vendas64.dtype, 'matriz')
## se tiver um elemento sendo texto todo o array vai ser do tipo texto
```

```python
# F065 Buscando elementos e filtrando valores em um array

arr65_a = np.arange(15).reshape(3,5)
print(arr65_a)
print(arr65_a[0])
print(arr65_a[0][1])
print(arr65_a[0][1:3])
print(arr65_a[1][arr65_a[1]>7])
print(arr65_a[1][(arr65_a[1]> 5) & (arr65_a[1] < 8) ])

print()
arr65_b = np.arange(15)
print(arr65_b)
print(arr65_b[0])
print(arr65_b[0:2])
print(arr65_b[arr65_b>7])

```

```python
# F066  Realizando operações com arrays (soma, média, mediana, potência, etc)

arr66 = np.arange(6).reshape(2,3)

print(arr66)

## Soma

print(arr66.sum())
print(arr66[0].sum())
print()
## Soma as linhas
print(arr66.sum(axis=1))
print()
## Soma acumulada
print(arr66.cumsum(axis=1))
print()

## Soma das colunas
print(arr66.sum(axis=0))
print(arr66.cumsum(axis=0))
print()

## Somar com numeros
print(arr66 + 1)
print()

## Multiplicar
print(arr66 * 2)
print()

## Menor valor da lista
print(arr66.min(), 'menor lista')
print(arr66.min(axis=1), 'menor linha') # menor linha
print(arr66.min(axis=0), 'Menor coluna') # menor coluna
print()

## Maior
print(arr66.max(), 'Maior da lsita')



```

```python

## Media

print(arr66.mean(), 'Media')
print(arr66.mean(axis=1), 'Media Linha')
print(arr66.mean(axis=0), 'Media Coluna')
print(arr66[0].mean(), 'Media Apenas da primeira linha')
print(arr66.mean(axis=1)[0], 'Media Apenas da primeira linha')
print(arr66.mean(axis=0)[0], 'Media primeira Coluna')

print(np.median(arr66), 'Mediana')
print(np.median(arr66[0]), 'Mediana')
print(np.median(arr66[1]), 'Mediana')

## Soma Array

print(arr66);
print(arr66[0] + arr66[1], 'soma array')
## Multiplica Array

print(arr66[0] * arr66[1], 'multiplica array')
print(arr66[0] ** arr66[1], 'potencia array')



```
