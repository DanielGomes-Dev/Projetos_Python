# F099 Criando um scatter plot usando apenas a documentação

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# make the data
np.random.seed(3)
x = 4 + np.random.normal(0, 2, 24)
y = 4 + np.random.normal(0, 2, len(x))
# size and color:
sizes = np.random.uniform(15, 80, len(x))
colors = np.random.uniform(15, 80, len(x))

# plot
fig, ax = plt.subplots()

ax.scatter(x, y, s=sizes, c=colors, vmin=0, vmax=100)
# ax.scatter(x, y)

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 8), yticks=np.arange(1, 8))

plt.show()
```

```python
import pandas as pd

base = pd.read_excel('dados-instagram.xlsx')

base.head()
```

```python
def tipo_para_numero(tipo):
    if(tipo== 'Foto'):
        return 1
    elif(tipo== 'Vídeo'):
        return 2
    elif(tipo== 'Reels'):
        return 3
    elif(tipo== 'IGTV'):
        return 4
    
base['NrTipo'] = base.Tipo.apply(tipo_para_numero)
base.head()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.colors as mcolors

plt.style.use('_mpl-gallery')

# make the data
# np.random.seed(3)
x = base['Curtidas']
y = base['Comentários']
# size and color:
# sizes = np.random.uniform(15, 80, len(x))
colors = base['NrTipo']

# plot
fig, ax = plt.subplots(figsize=(15,5))

# Paleta de cores: escolha uma perceptível
cmap = plt.cm.get_cmap('viridis', len(base['NrTipo'].unique()))

# Normalizar os valores de NrTipo para o cmap
norm = mcolors.Normalize(vmin=colors.min(), vmax=colors.max())


# ax.scatter(x, y, s=sizes, c=colors, vmin=0, vmax=100)
scatter_plt = ax.scatter(x, y, c=colors, cmap=cmap, norm=norm)


ax.legend(handles=scatter_plt.legend_elements()[0], labels=['Foto', 'Vídeo', 'Reels', 'IGTV'])

plt.show()
```
