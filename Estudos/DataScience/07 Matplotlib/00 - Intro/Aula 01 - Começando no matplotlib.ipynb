```python

# Importando as bibliotecas: Matplotlib e Pandas #F091
import matplotlib.pyplot as plt # Estudar mais
import numpy as np
import pandas as pd 

# make data

x = np.linspace(0,10, 100)

y = 4 + 2 * np.sin(2 * x)

# plot

fig, ax = plt.subplots()

## Primeiro elemento figura
## Segundo elemento o ax

ax.plot(x,y, linewidth=2.0)
ax.set(xlim=(0,6))
plt.show()
# Limitar o range do grafico
# xlim, ylim ->
# xticks -> yticks
# varios graficos na mesma figura
```
